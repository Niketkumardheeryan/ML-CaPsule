```python
transaction = []
for i in range(0,7500):
  transaction.append([str(data.values[i,j]) for j in range(0,20)])
transaction
```

**apyori library to perform market basket analysis or association rule learning**

```python
from apyori import apriori

model = apriori(transactions = transaction, min_support = 0.003, min_confidence = 0.2, min_lift = 3, min_length = 2)
```

```python
results = list(model)
results
```
