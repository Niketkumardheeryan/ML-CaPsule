# 🚨 Crime Analytics & Spatial-Temporal Risk Intelligence

This notebook provides advanced analytics and visualization capabilities on crime datasets, including temporal trend analysis across time periods, spatial hotspot identification, category distribution analysis, Month-over-Month (MoM) and Year-over-Year (YoY) comparison, location risk scoring, and automated summary observations.

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import matplotlib.pyplot as plt
import seaborn as sns
import os
import sys

# Import module components
from crime_analytics import CrimeDataAnalyzer, HotspotDetector, RiskScorer, generate_automated_insights

print("Libraries and Crime Analytics modules imported successfully.")
```

## 1. Data Ingestion & Overview

```python
df = pd.read_csv("crime_dataset.csv")
df["timestamp"] = pd.to_datetime(df["timestamp"])
print(f"Total Crime Records: {len(df)}")
df.head()
```

## 2. Key Performance Indicators (KPIs) & Data Overview

```python
analyzer = CrimeDataAnalyzer(df)
kpis = analyzer.get_summary_kpis()
for key, val in kpis.items():
    print(f"{key.replace('_', ' ').title()}: {val}")
```

## 3. Temporal Trend & MoM / YoY Growth Analysis

```python
ts_monthly = analyzer.get_time_series_trends(freq="ME")
growth_df = analyzer.get_mom_yoy_comparison()

fig = px.line(ts_monthly, x="timestamp", y="total_incidents", title="Monthly Crime Volume Trend", markers=True)
fig.show()

display(growth_df.tail(12))
```

## 4. Crime Category Distribution & Frequency Analysis

```python
cat_dist = analyzer.get_category_distribution()
plt.figure(figsize=(10, 5))
sns.barplot(data=cat_dist, x="count", y="crime_type", palette="magma")
plt.title("Crime Category Volume Frequency")
plt.xlabel("Count")
plt.ylabel("Crime Category")
plt.show()

display(cat_dist)
```

## 5. Hotspot Identification & Spatial Heatmap

```python
detector = HotspotDetector(df)
hotspots = detector.get_district_hotspots()
display(hotspots)

fig_map = px.density_mapbox(
    df, lat="latitude", lon="longitude", z="severity_score",
    radius=12, center=dict(lat=df["latitude"].mean(), lon=df["longitude"].mean()),
    zoom=10, mapbox_style="open-street-map", title="Geographical Crime Density Heatmap"
)
fig_map.show()
```

## 6. Location Risk Score Engine

```python
scorer = RiskScorer(df)
risk_scores = scorer.calculate_district_risk_scores(recency_weight=0.3, volume_weight=0.4, severity_weight=0.3)
display(risk_scores)

plt.figure(figsize=(8, 4))
sns.barplot(data=risk_scores, x="risk_score", y="district", palette="Reds_r")
plt.title("District Composite Risk Score Index (0 - 100 Scale)")
plt.xlabel("Risk Score")
plt.ylabel("District")
plt.show()
```

## 7. Automated Summary Insights

```python
insights = generate_automated_insights(df)
print("=== AUTOMATED OBSERVATIONS & INSIGHTS ===\n")
for i, ins in enumerate(insights, 1):
    print(f"{i}. {ins}")
```
