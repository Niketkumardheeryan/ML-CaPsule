```python
from bs4 import BeautifulSoup
import requests as r
```

```python
headers = {'User-Agent': 'Mozilla/5.0'}
response = r.get('https://coreyms.com', headers=headers)
soup = BeautifulSoup(response.text, 'lxml') #parse as lxml
```

```python
soup                                    
```

```python
soup.title
```

```python
soup.prettify()       # well oriented 
```

```python
print(soup.prettify())             #looks pretty to use prettify()
```

```python
# The following scripts are blocked as per agent/bot detection to avoid attacks.
# However the this is a simple way to check scripts.

script = soup.script

scripts = soup.find_all('script')

try:
    print(script.prettify())
except:
    print("The file does have <script> tag")

print(f"Scripts : {scripts}")
```

```python
# finfding a particular tags

print(f" with tag : {soup.title}")                                  #with tag          , will only get the first one title we create 
print(f"without tag : {soup.title.text}")                             #without tag
```

```python
link=soup.find('link')
```

```python
# splitting the link from '/ 'symbol 
link.split('/')[4]

#this is also a cause of cloudfare protection rules but this still can be used on simple websites!
```

```python
# search a particular tag we needed

style=soup.find('style')
print(style)
```

## Nested tags find 

```python
## for nested tags use this method

nested=soup.find('div',class_='site-container')

print(nested.find('g',class_='site-avatar-foreground').prettify())
```

```python
soup.find('style')                           # find() return a list 
```

## To find all tags with same names

```python
# for Looping over a articles or tag

for i in soup.find_all('link'):
    print(i)
    
                                                                       # all link  tags  will be printed
    
```
