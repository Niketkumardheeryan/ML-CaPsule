# Matplotlib Basics

Matplotlib is one of the most popular Python libraries for data visualization.

It helps us create:
- line plots
- bar charts
- scatter plots
- histograms
- boxplots
- subplots

In machine learning, Matplotlib is useful for understanding data before training a model.

```python
#import the library 

import matplotlib.pyplot as plt
import numpy as np
```

## 1. Simple Line Plot

A line plot is used to show how values change over a sequence.

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)
plt.show()
```

## 2. Adding Title and Axis Labels

A good plot should always tell a story clearly.

```python
plt.plot(x, y)
plt.title("Simple Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

## 3. Using Markers

Markers help us highlight individual data points.

```python
plt.plot(x, y, marker='o')
plt.title("Line Plot with Markers")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

## 4. Changing Line Style and Color

Matplotlib allows us to customize plots easily.

```python
plt.plot(x, y, color='red', linestyle='--', marker='o')
plt.title("Styled Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

## 5. Plotting Multiple Lines

We can compare more than one dataset in the same figure.

```python
x = np.array([1, 2, 3, 4, 5])
y1 = x * 2
y2 = x ** 2

plt.plot(x, y1, label='2x', marker='o')
plt.plot(x, y2, label='x^2', marker='s')
plt.title("Multiple Lines")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()
```

## 6. Changing Figure Size

Figure size helps make the plot more readable.

```python
plt.figure(figsize=(8, 4))
plt.plot(x, y2, marker='o')
plt.title("Figure with Custom Size")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

## 7. Adding Grid Lines

Grid lines make it easier to read values from the plot.

```python
plt.figure(figsize=(8, 4))
plt.plot(x, y2, marker='o')
plt.title("Plot with Grid")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.show()
```

## 8. Setting Axis Limits

We can control how much of the graph is visible.

```python
plt.figure(figsize=(8, 4))
plt.plot(x, y2, marker='o')
plt.title("Plot with Axis Limits")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.xlim(0, 6)
plt.ylim(0, 30)
plt.grid(True)
plt.show()
```

## 9. Bar Chart

Bar charts are useful for comparing categories.

```python
categories = ['A', 'B', 'C', 'D']
values = [10, 15, 7, 12]

plt.bar(categories, values)
plt.title("Simple Bar Chart")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

## 10. Horizontal Bar Chart

This is useful when category names are long.

```python
plt.barh(categories, values)
plt.title("Horizontal Bar Chart")
plt.xlabel("Values")
plt.ylabel("Categories")
plt.show()
```

## 11. Scatter Plot

Scatter plots are used to show the relationship between two variables.

```python
x = [1, 2, 3, 4, 5, 6]
y = [2, 4, 3, 7, 5, 8]

plt.scatter(x, y)
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

## 12. Histogram

Histograms show the distribution of numerical data.

```python
data = [12, 15, 16, 17, 18, 18, 19, 20, 21, 21, 22, 23, 25, 30, 31, 32]

plt.hist(data, bins=5, edgecolor='black')
plt.title("Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

## 13. Boxplot

Boxplots help us understand spread, median, and outliers.

```python
data1 = [10, 12, 14, 15, 18, 20, 22]
data2 = [8, 9, 10, 11, 13, 14, 16]

plt.boxplot([data1, data2], labels=['Group 1', 'Group 2'])
plt.title("Boxplot")
plt.ylabel("Values")
plt.show()
```

## 14. Pie Chart

Pie charts show parts of a whole.

```python
sizes = [40, 30, 20, 10]
labels = ['Python', 'Java', 'C++', 'JavaScript']

plt.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.title("Pie Chart")
plt.show()
```

## 15. Subplots

Subplots let us show multiple graphs in one figure.

```python
x = np.array([1, 2, 3, 4, 5])

fig, ax = plt.subplots(1, 2, figsize=(10, 4))

ax[0].plot(x, x**2, marker='o')
ax[0].set_title("x^2")

ax[1].plot(x, x**3, marker='o', color='green')
ax[1].set_title("x^3")

plt.tight_layout()
plt.show()
```

## 16. Customizing Ticks

Ticks help control what values appear on the axes.

```python
plt.plot(x, x**2)
plt.title("Custom Ticks")
plt.xticks([1, 2, 3, 4, 5])
plt.yticks([0, 5, 10, 15, 20, 25])
plt.grid(True)
plt.show()
```

## 17. Saving a Plot

Matplotlib plots can be saved as image files.

```python
plt.figure(figsize=(6, 4))
plt.plot(x, x**2, marker='o')
plt.title("Saved Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)

plt.savefig("my_plot.png", dpi=300, bbox_inches='tight')
plt.show()
```

## 18. Using Built-in Styles

Matplotlib has built-in styles that improve the look of plots quickly.

```python
plt.style.use('ggplot')

plt.figure(figsize=(6, 4))
plt.plot(x, x**2, marker='o')
plt.title("Styled Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

## 19. Resetting the Style

If you want the default style back, reset it.

```python
plt.style.use('default')
```
