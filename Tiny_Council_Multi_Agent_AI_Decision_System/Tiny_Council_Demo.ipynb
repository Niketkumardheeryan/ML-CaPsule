# 🏛️ Tiny-Council: Multi-Agent AI Decision System

**Tiny-Council** is a multi-agent AI decision-making framework where specialized AI agents (Analyst, Critic, Strategist, Ethics Evaluator, Synthesizer) collaborate, critique peer proposals, refine standpoints, and synthesize consensus decisions using structured voting algorithms and explainable reasoning traces.

---

```python
import sys
from pathlib import Path

# Ensure tiny_council package is accessible
sys.path.insert(0, str(Path('.').resolve()))

from tiny_council import (
    CouncilOrchestrator,
    ConsensusMode,
    AnalystAgent,
    CriticAgent,
    StrategistAgent,
    EthicsAgent,
    CustomAgent,
    ConsensusEngine,
    BordaCount,
    WeightedScoring,
    export_report_markdown,
    plot_consensus_trajectory,
)
print('Tiny-Council package successfully imported!')
```

## 1. Quick Start: Running a Decision Council

We evaluate a real-world technical architecture decision: **"Should our enterprise platform transition from a monolith to a microservices architecture?"**

```python
orchestrator = CouncilOrchestrator()
topic = 'Should our enterprise platform transition from a monolith to a microservices architecture?'

context = orchestrator.run_council(
    topic=topic,
    rounds=2,
    mode=ConsensusMode.DELPHI_METHOD
)

print(f'Execution Time: {context.get_duration()} seconds')
print(f'Final Agreement Index: {context.consensus_summary["final_agreement_index"]}')
print(f'Disagreement Entropy: {context.consensus_summary["disagreement_entropy"]}')
```

## 2. Inspecting Executive Resolution & Agent Standpoints

```python
print('=== EXECUTIVE RESOLUTION ===\n')
print(context.executive_resolution)
print('\n' + '='*50 + '\n')
print('=== INDIVIDUAL AGENT POSITIONS ===\n')
for proposal in context.refined_proposals:
    print(f'Agent: {proposal["agent_name"]} ({proposal["role"]})')
    print(f'Confidence Score: {proposal["confidence"]} | Weight: {proposal["weight"]}')
    print(proposal.get('refined_response') or proposal.get('response'))
    print('-'*40)
```

## 3. Mathematical Consensus Algorithms & Voting

Tiny-Council computes Borda Count ranks and weighted scoring metrics.

```python
# Example Borda Count ranking across candidate options
options = ['Option A: Full Microservices', 'Option B: Modular Monolith', 'Option C: Serverless Hybrids']
rankings = [
    ['Option B: Modular Monolith', 'Option A: Full Microservices', 'Option C: Serverless Hybrids'], # Analyst
    ['Option B: Modular Monolith', 'Option C: Serverless Hybrids', 'Option A: Full Microservices'], # Critic
    ['Option A: Full Microservices', 'Option B: Modular Monolith', 'Option C: Serverless Hybrids'], # Strategist
    ['Option B: Modular Monolith', 'Option C: Serverless Hybrids', 'Option A: Full Microservices'], # Ethics
]
weights = [1.2, 1.0, 1.1, 1.0]

borda_res = BordaCount.evaluate(options, rankings, weights)
print('Borda Count Voting Outcome:')
print(f'Consensus Winner: {borda_res["winner"]}')
print('Scores:', borda_res['scores'])
```
