## What is Prompt Engineering?

A **prompt** is the text you send to an AI model (like ChatGPT, Gemini, or Claude) to get a response.  
**Prompt Engineering** is the art and science of writing prompts that get you *exactly* the output you want.

Think of it like this: the same question asked in different ways can give you wildly different answers.  
Prompt engineering teaches you how to ask the *right* way.

---

## Why Does It Matter?

- 🔍 Better prompts = better AI outputs
- 💼 Prompt Engineering is one of the **most in-demand AI skills in 2025–2026**
- 🆓 You don't need to train a model — just learn to *talk* to one well
- 🚀 Used in production by companies to build AI products, automate workflows, and generate content

````python
# ============================================================
# CELL 2: Setup — Import libraries and enter your API key
# ============================================================

import os
from getpass import getpass
from groq import Groq

# Securely enter your Groq API key (it won't show on screen)
api_key = getpass("Enter your Groq API key: ")

# Initialize the Groq client
client = Groq(api_key=api_key)

# We'll use Llama 3.1 8B — fast, free, and very capable for learning
MODEL = "llama-3.1-8b-instant"

print("✅ Setup complete! You're ready to start prompt engineering.")
````

````python
# ============================================================
# CELL 3: Helper function — used throughout this notebook
# This sends a prompt to the LLM and returns the response.
# ============================================================

def ask_llm(prompt: str, system: str = None, temperature: float = 0.7) -> str:
    """
    Send a prompt to the Groq LLM and return its response as a string.

    Args:
        prompt      : The user message / prompt text
        system      : Optional system message (used for role-based prompting)
        temperature : Controls randomness. 0 = deterministic, 1 = creative.

    Returns:
        The model's response as a plain string.
    """
    messages = []

    if system:
        messages.append({"role": "system", "content": system})

    messages.append({"role": "user", "content": prompt})

    response = client.chat.completions.create(
        model=MODEL,
        messages=messages,
        temperature=temperature,
        max_tokens=1024,
    )

    return response.choices[0].message.content


def display_result(technique: str, prompt: str, response: str):
    """Pretty-print the prompt and model response."""
    print(f"{'='*60}")
    print(f"🔧 Technique : {technique}")
    print(f"{'='*60}")
    print(f"📝 PROMPT:\n{prompt}")
    print(f"\n{'─'*60}")
    print(f"🤖 MODEL RESPONSE:\n{response}")
    print(f"{'='*60}\n")


print("✅ Helper functions loaded.")
````

# Technique 1: Zero-Shot Prompting

## What is it?

**Zero-shot** means you ask the model to do something *without* giving it any examples.  
You just describe the task directly and trust the model's existing knowledge.

## When to use it?
- For straightforward tasks the model clearly understands (summarization, translation, Q&A)
- When you want a quick answer and don't need high precision
- As a baseline before trying more advanced techniques

## How it works

```
[Task Description] → Model → Response
```

No examples. No guidance on *how* to answer. Just a direct question or instruction.

## 💡 Pro Tips
- Be **specific** about what you want. "Summarize this" is worse than "Summarize this in 3 bullet points for a non-technical audience."
- Use **action verbs**: Classify, Summarize, Translate, Extract, List, Explain
- Specify the **format** if you care about it (e.g., "in one sentence", "as a list")

````python
# ============================================================
# TECHNIQUE 1: Zero-Shot Prompting
# ============================================================

# --- Example 1a: Bad zero-shot (vague prompt) ---
bad_prompt = "Tell me about climate change."

# --- Example 1b: Good zero-shot (specific, structured) ---
good_prompt = """
Explain climate change in exactly 3 bullet points.
Each bullet should be one sentence and use simple language 
suitable for a 10-year-old.
"""

# --- Example 1c: Practical task — Sentiment Classification ---
classification_prompt = """
Classify the sentiment of the following customer review as 
POSITIVE, NEGATIVE, or NEUTRAL. Reply with only one word.

Review: "The delivery was late but the product quality exceeded my expectations."
"""

print("Running zero-shot examples...\n")

# Run bad vs good comparison
bad_response  = ask_llm(bad_prompt)
good_response = ask_llm(good_prompt)
clf_response  = ask_llm(classification_prompt, temperature=0.0)  # 0 = deterministic for classification

print("❌ VAGUE PROMPT OUTPUT:")
print(bad_response[:300], "...\n")  # Truncate to show it's long and unfocused

display_result("Zero-Shot (Specific)", good_prompt.strip(), good_response)
display_result("Zero-Shot (Classification)", classification_prompt.strip(), clf_response)
````

````python
# ============================================================
# EXPERIMENT: Try your own zero-shot prompt!
# ============================================================

# 👇 Change the task below and run this cell
your_task = """
Translate the following sentence to French in a formal tone:
"I would like to schedule a meeting for next Monday afternoon."
"""

response = ask_llm(your_task)
display_result("Zero-Shot (Your Experiment)", your_task.strip(), response)
````

### 📊 What We Observed

| Prompt Type | Output Quality |
|-------------|---------------|
| Vague ("Tell me about X") | Long, generic, unfocused |
| Specific with format constraints | Concise, structured, on-point |
| Task-specific with output format | Precise, usable immediately |

> **Key Takeaway:** The more specific your zero-shot prompt, the less the model has to "guess" what you want. Constraints are your friend.

# Technique 2: Few-Shot Prompting

## What is it?

**Few-shot prompting** means providing a few examples ("shots") of input→output pairs *before* your actual question.  
The model learns the pattern from your examples and applies it to new inputs.

## When to use it?
- When the task has a specific format or style you need the model to follow
- When zero-shot gives inconsistent or wrong outputs
- For classification, transformation, or extraction tasks
- When you have a custom schema the model doesn't know about

## How it works

```
Example 1: Input → Output
Example 2: Input → Output  
Example 3: Input → Output
Now answer: New Input → ?
```

## 💡 Pro Tips
- Use **2–5 examples** — more isn't always better, and it wastes tokens
- Make examples **diverse** — cover different cases, not just easy ones
- Keep examples **consistent** in format — the model copies your structure
- Put the **actual question last**, after all examples

````python
# ============================================================
# TECHNIQUE 2: Few-Shot Prompting
# ============================================================

# --- Example 2a: Sentiment with custom label schema ---
# Imagine your system uses labels: [HAPPY, SAD, ANGRY, NEUTRAL]
# Zero-shot won't know this. Few-shot teaches it.

few_shot_sentiment = """
Classify the emotion in each sentence using exactly one label 
from: [HAPPY, SAD, ANGRY, NEUTRAL]

Examples:
Sentence: "I just got promoted at work!" → HAPPY
Sentence: "The flight got cancelled and now I'm stuck here." → ANGRY  
Sentence: "My dog passed away yesterday." → SAD
Sentence: "The meeting is at 3pm." → NEUTRAL

Now classify:
Sentence: "I can't believe they ran out of tickets — I waited in line for 4 hours!"
"""

# --- Example 2b: Structured data extraction ---
# Teaching the model your exact extraction format
few_shot_extraction = """
Extract the name, role, and company from each sentence. 
Format: Name | Role | Company

Examples:
"Sundar Pichai, the CEO of Google, announced the new AI features."
→ Sundar Pichai | CEO | Google

"Indra Nooyi served as the Chairperson of PepsiCo for over a decade."
→ Indra Nooyi | Chairperson | PepsiCo

"Sam Altman is the Chief Executive Officer at OpenAI."
→ Sam Altman | Chief Executive Officer | OpenAI

Now extract:
"Satya Nadella, Microsoft's CEO, unveiled the company's latest AI roadmap."
"""

# --- Example 2c: Style transfer (rewrite in a specific tone) ---
few_shot_style = """
Rewrite the sentence in a formal, professional business email tone.

Examples:
Casual: "Hey, can we chat about the project sometime?"
Formal: "I would appreciate the opportunity to discuss the project at your earliest convenience."

Casual: "The client is super unhappy with the delay."
Formal: "The client has expressed significant dissatisfaction regarding the project timeline."

Now rewrite:
Casual: "We totally dropped the ball on this one, sorry!"
"""

print("Running few-shot examples...\n")

r1 = ask_llm(few_shot_sentiment, temperature=0.0)
r2 = ask_llm(few_shot_extraction, temperature=0.0)
r3 = ask_llm(few_shot_style, temperature=0.3)

display_result("Few-Shot (Emotion Classification)", few_shot_sentiment.strip(), r1)
display_result("Few-Shot (Entity Extraction)", few_shot_extraction.strip(), r2)
display_result("Few-Shot (Style Transfer)", few_shot_style.strip(), r3)
````

````python
# ============================================================
# Zero-Shot vs Few-Shot: Direct Comparison
# ============================================================
# Let's see the actual difference for the SAME task

task_description = """
Classify the urgency of the following IT support ticket.
Use the label: CRITICAL, HIGH, MEDIUM, or LOW.

Ticket: "I can't log into the company VPN. Working from home and have a client 
call in 20 minutes."
"""

zero_shot_response = ask_llm(task_description, temperature=0.0)

few_shot_version = """
Classify the urgency of IT support tickets.
Labels: CRITICAL (system-wide outage), HIGH (individual blocked on deadline), 
MEDIUM (inconvenient but workaround exists), LOW (cosmetic/nice-to-have)

Examples:
Ticket: "The entire production server is down." → CRITICAL
Ticket: "My keyboard's spacebar is a bit sticky." → LOW
Ticket: "Can't access Slack on my phone, using desktop for now." → MEDIUM
Ticket: "My laptop won't boot and I have a board presentation in 1 hour." → CRITICAL

Now classify:
Ticket: "I can't log into the company VPN. Working from home and have a client 
call in 20 minutes."
"""

few_shot_response = ask_llm(few_shot_version, temperature=0.0)

print("🔬 COMPARISON: Same task, different prompting strategies\n")
print(f"Zero-Shot Output: {zero_shot_response}")
print(f"Few-Shot Output:  {few_shot_response}")
print("\n💡 Notice how few-shot anchors the model to YOUR definition of each label.")
````

### 📊 Zero-Shot vs Few-Shot

| | Zero-Shot | Few-Shot |
|--|-----------|----------|
| Examples given | None | 2–5 |
| Best for | General tasks | Custom schemas, formats |
| Token cost | Low | Medium |
| Consistency | Lower | Higher |
| When to use | Simple, well-known tasks | Domain-specific, custom output format |

> **Key Takeaway:** Few-shot is how you "teach" the model *your* format without fine-tuning. It's incredibly powerful for production use cases.

# Technique 3: Chain-of-Thought (CoT) Prompting

## What is it?

**Chain-of-Thought prompting** instructs the model to show its reasoning step by step before giving a final answer.  
Instead of jumping to a conclusion, the model "thinks out loud."

Research from Google (Wei et al., 2022) showed that adding *"Let's think step by step"* to math problems dramatically improved accuracy — even on problems the model got wrong before.

## When to use it?
- Math problems and calculations
- Multi-step logical reasoning
- Complex decision-making
- Debugging or root-cause analysis
- Any task where the answer depends on intermediate steps

## Two styles of CoT

**Style 1 — Zero-Shot CoT:** Just add *"Think step by step"* to your prompt  
**Style 2 — Few-Shot CoT:** Provide examples that show the reasoning chain

## 💡 Pro Tips
- Use `temperature=0` or low temperature for reasoning tasks — you want logic, not creativity
- Ask the model to put the final answer on a **separate labeled line** — makes it easy to parse
- For very complex problems, you can chain multiple CoT calls (each output becomes the next input)

````python
# ============================================================
# TECHNIQUE 3: Chain-of-Thought (CoT) Prompting
# ============================================================

# --- Example 3a: Math Problem — No CoT vs CoT ---
math_problem = """
A store sells apples at ₹15 each and oranges at ₹12 each.
Priya buys 8 apples and 5 oranges. She pays with a ₹300 note.
How much change does she receive?
"""

# Without CoT — direct answer
no_cot_prompt  = math_problem + "\nAnswer:"

# With Zero-Shot CoT — just add the magic phrase
zero_shot_cot  = math_problem + "\nThink step by step and show your work. Then give the final answer."

# --- Example 3b: Few-Shot CoT — teach the reasoning pattern ---
few_shot_cot = """
Solve the following word problem. Show your reasoning at each step, 
then state the final answer on a line that starts with "Final Answer:"

Example:
Problem: A train travels at 60 km/h for 2 hours, then 80 km/h for 1.5 hours. 
         What is the total distance?
Reasoning:
  Step 1: Distance in first leg = Speed × Time = 60 × 2 = 120 km
  Step 2: Distance in second leg = 80 × 1.5 = 120 km  
  Step 3: Total distance = 120 + 120 = 240 km
Final Answer: 240 km

Problem: A company had 500 employees. They hired 20% more in Q1, then laid off 
         50 people in Q2. How many employees do they have now?
"""

# --- Example 3c: Logical reasoning with CoT ---
logic_cot = """
Answer the following logic puzzle. Think through it step by step before answering.

Puzzle:
All machine learning engineers know Python.
Ananya knows Python.
Does it logically follow that Ananya is a machine learning engineer?

Reason through this carefully:
"""

print("Running Chain-of-Thought examples...\n")

r_no_cot       = ask_llm(no_cot_prompt, temperature=0.0)
r_zero_cot     = ask_llm(zero_shot_cot, temperature=0.0)
r_few_shot_cot = ask_llm(few_shot_cot, temperature=0.0)
r_logic_cot    = ask_llm(logic_cot, temperature=0.0)

print("🔬 Without CoT:")
print(r_no_cot, "\n")
print("─" * 60)

display_result("Chain-of-Thought (Zero-Shot CoT)", zero_shot_cot.strip(), r_zero_cot)
display_result("Chain-of-Thought (Few-Shot CoT)", few_shot_cot.strip(), r_few_shot_cot)
display_result("Chain-of-Thought (Logic Puzzle)", logic_cot.strip(), r_logic_cot)
````

````python
# ============================================================
# CoT for Code Debugging — a real-world use case!
# ============================================================

debug_cot = """
The following Python function is supposed to return the second largest number 
in a list. However, it has a bug. 

```python
def second_largest(nums):
    nums.sort()
    return nums[-2]
```

Test case: second_largest([3, 3, 1]) should return 1, but this returns 3.

Step by step:
1. Trace through what the function does with [3, 3, 1]
2. Identify why the output is wrong
3. Write the corrected version of the function
4. Explain what the fix does differently
"""

r_debug = ask_llm(debug_cot, temperature=0.0)
display_result("CoT (Code Debugging)", debug_cot.strip(), r_debug)
````

### 📊 Impact of Chain-of-Thought

| Task Type | Without CoT | With CoT |
|-----------|-------------|----------|
| Simple math | Often correct | Correct + verifiable |
| Multi-step math | Frequently wrong | Much more accurate |
| Logical reasoning | May jump to wrong conclusion | Catches faulty logic |
| Code debugging | Spots the obvious | Traces through systematically |

> **Key Takeaway:** CoT prompting essentially forces the model to do the work it would otherwise skip. The reasoning chain also makes errors *visible* so you can catch them — a huge advantage over black-box answers.

# Technique 4: Role-Based Prompting

## What is it?

**Role-based prompting** assigns a specific persona, expertise, or character to the model before asking your question.  
By telling the model *who it is*, you dramatically shift the style, depth, and vocabulary of its responses.

This works because LLMs are trained on text from many different domains. By activating a specific domain ("you are a cardiologist"), you steer the model toward language patterns, vocabulary, and reasoning styles from that domain.

## When to use it?
- When you need domain-specific expertise (medical, legal, technical)
- When you need a specific communication style (formal, casual, ELI5)
- For building AI assistants with consistent personas
- When writing content for a specific audience

## How it works — The System Message

In chat-based LLMs, there are typically 3 roles:
- **System:** Persistent instructions and persona (set once at the start)
- **User:** Your actual question
- **Assistant:** The model's response

Role-based prompting usually goes in the **system message** — it stays active for the whole conversation.

## 💡 Pro Tips
- Be specific about the role: "expert Python developer with 10 years of experience" > "developer"
- Include **behavioral constraints** in the role: "always respond with code examples", "never use jargon"
- Combine role + task for maximum effect: Role defines WHO, task defines WHAT

````python
# ============================================================
# TECHNIQUE 4: Role-Based Prompting
# ============================================================

# The SAME question answered by 3 different "experts"
# This shows how powerfully the role shapes the response.

question = "What should I know about neural networks?"

# --- Role 1: Professor explaining to a 10-year-old ---
role_teacher = """
You are a friendly, enthusiastic professor who specializes in explaining 
complex AI concepts to children aged 8-12. You use simple analogies, 
fun comparisons, and relatable examples from everyday life. 
You never use technical jargon without immediately explaining it in simple words.
Keep your answer to 4-5 sentences.
"""

# --- Role 2: Senior ML Engineer mentoring a junior ---
role_engineer = """
You are a Senior Machine Learning Engineer at a top tech company with 8+ years 
of experience building production ML systems. You are mentoring a junior developer 
who knows Python but is new to ML. Your explanations are practical, code-focused, 
and always include real-world considerations like training time, overfitting, 
and deployment challenges.
"""

# --- Role 3: Technical interviewer ---
role_interviewer = """
You are a technical interviewer at a top AI company preparing a candidate 
for a machine learning engineer interview. Structure your answer as the 
key concepts they MUST know, what interviewers commonly ask, 
and one trick question to watch out for.
"""

print("Same question → 3 different roles → 3 completely different responses\n")
print(f"Question: '{question}'\n")
print("=" * 60)

r_teacher      = ask_llm(question, system=role_teacher, temperature=0.7)
r_engineer     = ask_llm(question, system=role_engineer, temperature=0.7)
r_interviewer  = ask_llm(question, system=role_interviewer, temperature=0.5)

print("\n👩‍🏫 ROLE: Friendly Professor (for kids)")
print("-" * 60)
print(r_teacher)

print("\n\n👨‍💻 ROLE: Senior ML Engineer (for junior devs)")
print("-" * 60)
print(r_engineer)

print("\n\n🎯 ROLE: Technical Interviewer")
print("-" * 60)
print(r_interviewer)
````

````python
# ============================================================
# Advanced: Role-Based Prompting for Building AI Assistants
# ============================================================
# This is how real products like customer support bots are built.

product_assistant_role = """
You are Aria, a helpful customer support assistant for CloudStore, 
an online electronics store.

Your personality:
- Friendly, professional, and patient
- Always empathetic when customers are frustrated
- Solution-focused — always try to resolve the issue

Your constraints:
- You can help with: orders, returns, product questions, shipping
- You CANNOT share: internal pricing data, other customers' info
- If you don't know something, say: "Let me connect you with a specialist for that."
- Always end your message by asking if there's anything else you can help with.

Store policies:
- Returns accepted within 30 days with original packaging
- Free shipping on orders above ₹999
- Delivery takes 3-5 business days
"""

customer_message = """
I ordered a laptop 8 days ago and it still hasn't arrived. 
I paid extra for fast delivery. I'm really frustrated because 
I needed it for work this week.
"""

r_assistant = ask_llm(customer_message, system=product_assistant_role, temperature=0.6)
display_result("Role-Based (Customer Support Bot)", customer_message.strip(), r_assistant)
````

### 📊 What Role-Based Prompting Changes

| Aspect | Without Role | With Role |
|--------|-------------|----------|
| Vocabulary | Generic | Domain-specific |
| Depth | Surface-level | Appropriate to expertise |
| Tone | Neutral | Matched to persona |
| Focus | Broad | Targeted |
| Use in products | ❌ | ✅ Core technique |

> **Key Takeaway:** The **system message is your control panel**. Every production AI application (chatbots, assistants, tutors) uses role-based prompting to maintain consistent persona and constrain model behavior.

# Technique 5: Structured Output Prompting

## What is it?

**Structured Output Prompting** tells the model to respond in a specific machine-readable format — most commonly **JSON**, but also Markdown tables, CSV, XML, or YAML.

This is **critical for production AI systems** because:
- Your code needs to *parse* the model's response programmatically
- Free-form text is hard to process reliably
- Structured output lets you plug AI directly into databases, APIs, and UIs

## When to use it?
- Data extraction from unstructured text
- Generating content that will be displayed in a UI
- Any scenario where another system will consume the model's output
- Building AI pipelines where one model's output feeds another

## 💡 Pro Tips
- Always provide the **exact JSON schema** you want — key names, data types
- Say **"respond ONLY with valid JSON, no extra text"** — prevents markdown fences and preambles
- Use `temperature=0` — deterministic responses are more format-stable
- Wrap the parse in try/except — models occasionally deviate from the schema
- For complex schemas, give an **example JSON** in the prompt

````python
# ============================================================
# TECHNIQUE 5: Structured Output Prompting
# ============================================================

import json

# --- Example 5a: Extract structured data from a job posting ---
job_posting = """
We're looking for a Machine Learning Engineer at TechCorp Bangalore.
The role requires 3+ years of experience with PyTorch and TensorFlow.
Knowledge of MLOps, Docker, and Kubernetes is a plus.
CTC offered: 18-25 LPA depending on experience.
Apply by: December 31, 2025. 
Contact: careers@techcorp.in
"""

extraction_prompt = f"""
Extract information from the job posting below and return it as a JSON object.
Use EXACTLY this schema:
{{
  "job_title": "string",
  "company": "string",
  "location": "string",
  "experience_required_years": "number",
  "required_skills": ["list of strings"],
  "nice_to_have_skills": ["list of strings"],
  "salary_range_lpa": {{"min": "number", "max": "number"}},
  "application_deadline": "string (YYYY-MM-DD format)",
  "contact_email": "string"
}}

Respond ONLY with the JSON object. No markdown, no explanation, no extra text.

Job Posting:
{job_posting}
"""

raw_json = ask_llm(extraction_prompt, temperature=0.0)

# Parse and pretty-print the JSON
try:
    # Strip markdown fences if model adds them (safety net)
    clean_json = raw_json.strip().strip("```json").strip("```").strip()
    parsed = json.loads(clean_json)
    print("✅ Successfully extracted structured data:\n")
    print(json.dumps(parsed, indent=2))
    
    # Now you can access it programmatically!
    print(f"\n📌 Parsed fields:")
    print(f"   Job Title : {parsed['job_title']}")
    print(f"   Salary    : {parsed['salary_range_lpa']['min']}–{parsed['salary_range_lpa']['max']} LPA")
    print(f"   Skills    : {', '.join(parsed['required_skills'])}")
except json.JSONDecodeError as e:
    print(f"⚠️ Parsing failed: {e}")
    print("Raw output:", raw_json)
````

````python
# --- Example 5b: Generate a Markdown table ---
# Useful when output will be displayed in a README or documentation

table_prompt = """
Compare the following 4 ML algorithms across these dimensions:
- Algorithm Name
- Type (Supervised/Unsupervised/Both)
- Best Used For
- Handles Non-linear Data (Yes/No)
- Interpretability (Low/Medium/High)

Algorithms: Linear Regression, Decision Tree, K-Means, Random Forest

Return ONLY a properly formatted Markdown table. No introduction, no conclusion.
"""

table_response = ask_llm(table_prompt, temperature=0.0)
print("📊 Markdown Table Output:")
print(table_response)
````

````python
# --- Example 5c: Generate a list of objects as JSON array ---
# A very common pattern: generate content at scale

flashcard_prompt = """
Generate 5 flashcards for learning Python list comprehensions.
Return a JSON array where each element has:
{
  "question": "the question or concept",
  "answer": "the concise answer",
  "example": "a one-line Python code example",
  "difficulty": "beginner | intermediate | advanced"
}

Respond with ONLY the JSON array. No markdown, no text outside the JSON.
"""

flashcard_response = ask_llm(flashcard_prompt, temperature=0.3)

try:
    clean = flashcard_response.strip().strip("```json").strip("```").strip()
    flashcards = json.loads(clean)
    
    print(f"✅ Generated {len(flashcards)} flashcards:\n")
    for i, card in enumerate(flashcards, 1):
        print(f"── Card {i} [{card['difficulty'].upper()}] ──")
        print(f"   Q: {card['question']}")
        print(f"   A: {card['answer']}")
        print(f"   Example: {card['example']}\n")
        
    print("💡 These are now Python dictionaries — you could save them to a database,")
    print("   display them in a web app, or export them to a flashcard app!")
except json.JSONDecodeError as e:
    print(f"⚠️ Parsing failed: {e}")
    print(flashcard_response)
````

### 📊 Structured Output Summary

| Format | Use Case | Parse With |
|--------|----------|------------|
| JSON Object | Single entity extraction | `json.loads()` |
| JSON Array | Generating multiple items | `json.loads()` |
| Markdown Table | Documentation, READMEs | Display directly |
| CSV | Tabular data for spreadsheets | `csv` module |
| YAML | Config files, pipelines | `yaml.safe_load()` |

> **Key Takeaway:** Structured outputs are what transform an LLM from a "chatbot" into a **programmable AI component**. This is the technique that powers most real-world AI integrations.

# 🔁 Combining Techniques — The Real Power

In practice, the best prompts **combine multiple techniques**. Let's see a real-world example that uses all 5 at once.

````python
# ============================================================
# COMBINED: All 5 techniques in one prompt
# Use case: AI-powered resume screener
# ============================================================

# Role-Based: Expert recruiter
recruiter_role = """
You are a senior technical recruiter at a top AI company with 10 years of experience 
screening ML engineer candidates. You are thorough, objective, and always provide 
actionable feedback. You evaluate candidates fairly regardless of their background.
"""

candidate_info = """
Candidate: Priya Sharma
Degree: B.Tech CSE (AI/ML), CGPA 8.7, 2nd Year
Skills: Python, TensorFlow, SQL, FastAPI, Docker
Projects: Built a sentiment analysis API deployed on AWS Lambda, 
          Contributed to 2 open-source ML repos (150 GitHub stars combined)
Research: 1 paper under review at a medical journal on EEG signal classification
Hackathons: Won 2 national-level hackathons in 2024
"""

job_requirements = """
Role: ML Engineering Intern
Requirements: Python, ML fundamentals, ability to build and deploy models, 
              some project experience. PhD not required.
"""

# Few-Shot: Show evaluation format + Chain-of-Thought: reason before scoring
# Structured Output: JSON
combined_prompt = f"""
You will evaluate a candidate for an ML internship.

EXAMPLE EVALUATION (use this format):
Candidate: Basic background, 1 project, no research.
Reasoning:
  - Has core Python skills: YES, basic level
  - ML project experience: YES but limited
  - Research signal: NONE
  - Differentiators: none notable
Result: {{"score": 5, "recommendation": "MAYBE", "key_strengths": ["Python basics"], 
          "gaps": ["No research", "Limited project depth"], 
          "suggested_interview_focus": "Assess ML fundamentals depth"}}

Now evaluate this candidate. Think step by step, then output ONLY the JSON result.

Job Requirements:
{job_requirements}

Candidate Profile:
{candidate_info}

Think through each requirement, then output the JSON.
"""

combined_result = ask_llm(combined_prompt, system=recruiter_role, temperature=0.2)

print("🎯 Combined Prompt Engineering — Resume Screener\n")
print("Techniques used: Role-Based + Few-Shot + Chain-of-Thought + Structured Output\n")
print("─" * 60)
print(combined_result)

# Try to parse the JSON from the response
import re
json_match = re.search(r'\{[\s\S]+\}', combined_result)
if json_match:
    try:
        evaluation = json.loads(json_match.group())
        print("\n✅ Parsed evaluation:")
        print(f"   Score          : {evaluation.get('score')}/10")
        print(f"   Recommendation : {evaluation.get('recommendation')}")
        print(f"   Key Strengths  : {evaluation.get('key_strengths')}")
    except:
        pass
````

# 📋 Quick Reference: Prompt Engineering Cheat Sheet

## The 5 Techniques

| Technique | Trigger Phrase | Best For |
|-----------|----------------|----------|
| **Zero-Shot** | Direct question with constraints | General tasks, baselines |
| **Few-Shot** | "Examples: ... Now:" | Custom formats, domain-specific labels |
| **Chain-of-Thought** | "Think step by step" | Math, logic, multi-step reasoning |
| **Role-Based** | System message with persona | Style control, expertise, assistants |
| **Structured Output** | "Respond only with JSON" | Data extraction, AI pipelines |

## Universal Best Practices

```
✅ DO:
  - Be specific about format ("in 3 bullet points", "as JSON", "in one sentence")
  - Specify the audience ("explain to a 10-year-old", "for a senior engineer")
  - Set constraints ("maximum 100 words", "only use terms from this list")
  - Use low temperature (0.0–0.3) for factual/structured tasks
  - Use higher temperature (0.7–1.0) for creative tasks
  - Test and iterate — prompting is empirical

❌ DON'T:
  - Ask vague questions expecting specific answers
  - Use negative instructions alone ("don't be verbose") — add positive ones too
  - Assume one prompt works for all models — test on your target model
  - Skip validation — always handle unexpected output formats in code
```

## Temperature Guide

| Task | Temperature |
|------|-------------|
| Classification, extraction | 0.0 |
| Reasoning, math | 0.0–0.2 |
| Technical writing | 0.3–0.5 |
| Conversational, summaries | 0.5–0.7 |
| Creative writing, brainstorming | 0.7–1.0 |
